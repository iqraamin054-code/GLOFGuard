## Load the raw data

```python
import pandas as pd

df = pd.read_csv("../data/raw/pakistan_full_training_dataset.csv")
print(df.shape)
df.head()
```

## Check for missing values and duplicates

```python
print(df.isnull().sum())

print("Duplicates:" , df.duplicated().sum())
```

## Drop the unnecessary columns

```python
df_clean = df.drop(columns=['OBJECTID_1', 'Id', 'OBJECTID', 'Shape_Leng', 'Shape_Le_1', 'Shape_Area'])
print(df_clean.shape)
df_clean.columns
```

## Check basic stats

```python
df_clean.describe()
```

## SAVE CLEAN DATA

```python
df_clean.to_csv("../data/processed/pakistan_full_training_dataset_clean.csv", index=False)
```
